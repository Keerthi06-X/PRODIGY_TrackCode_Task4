```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/sample_data/twitter_training.csv.zip", skiprows=4)
df.head()
```

```python
print(df.shape)
print(df.info())
print(df.isnull().sum())
```

```python
#Removing Mssing values
df = df.dropna()
print(df.isnull().sum())
```

```python
#sentiment count
print(df['Positive'].value_counts())
```

```python
plt.figure(figsize=(6,5))
sns.countplot(x='Positive', data=df)
plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Count")

plt.show()
```

```python
sentiment_counts = df['Positive'].value_counts()
plt.figure(figsize=(4,4))
plt.pie(
    sentiment_counts,
    labels=sentiment_counts.index,
    autopct='%1.1f%%'
)
plt.title("Sentiment Percentage Distribution")

plt.show()
```

```python
top_brands = df['Borderlands'].value_counts().head(10)
plt.figure(figsize=(8,4))
top_brands.plot(kind='bar')
plt.title("Top Discussed Brands")
plt.xlabel("Brand")
plt.ylabel("Mentions")

plt.show()
```
